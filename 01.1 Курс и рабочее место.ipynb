## Git: машина времени для кода

Представьте: вы неделю писали проект, что-то поменяли — и всё сломалось. Как вернуть вчерашнюю версию? Заводить папки `проект_финал`, `проект_финал_2`, `проект_точно_финал`? Знакомо?

**Git** запоминает состояния проекта. Каждое сохранённое состояние называется **коммит**, и к любому можно вернуться. Git хранит историю локально, а GitHub позволяет держать копию в сети и показывать результат. До отправки на другое устройство один локальный коммит не заменяет резервную копию.

Забавный факт: Git написал Линус Торвальдс в 2005 году, когда ему надоели существующие системы. Слово «git» в британском сленге означает «неприятный тип» — Торвальдс шутил, что называет проекты в честь себя.

```python
import shutil
import subprocess

if shutil.which('git'):
    print(subprocess.run(['git', '--version'], capture_output=True, text=True, timeout=10).stdout.strip())
else:
    print('Git пока не установлен. Сегодня работаем с файлами, настройку оставляем на неделю.')
```

Главные команды выполняем в терминале; шпаргалка — `docs/GIT.md` своего проекта.

| Команда | Что делает |
|---|---|
| `git clone <адрес>` | скачать репозиторий |
| `git pull` | получить обновления в его локальной копии |
| `git status` | увидеть изменения |
| `git add <файл>` | выбрать файлы для коммита |
| `git commit -m "что сделал"` | сохранить состояние |
| `git push` | отправить коммиты в свой GitHub |
