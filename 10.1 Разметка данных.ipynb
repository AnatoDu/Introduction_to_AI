`item_id` — устойчивый ключ обращения, `annotator` — условное обозначение разметчика, не его ФИО.
Разделяем исходный текст и ответы. Одна строка ответа — один объект, один разметчик, одна версия правила.
Для примера возьмём заранее **вымышленные** ответы A/B. Они нужны для работающего общего кода,
а не для подмены ваших попыток. В основном примере не используем переменные из «Ваш ход».

`to_csv(..., index=False)` сохраняет только столбцы, без дополнительного номера строки.
Для собственных ответов используйте отдельное имя файла. Создание пустой формы ниже не затирает готовую.


```python
from pathlib import Path

output_dir = WORKSPACE_DIR / 'outputs' / 'annotation'
output_dir.mkdir(parents=True, exist_ok=True)
form_path = output_dir / 'my_annotations.csv'
if not form_path.exists():
    form = messages[['item_id']].copy()
    form['annotator'] = 'my_alias'
    form['label'] = ''
    form['guideline_version'] = 'v1'
    form['note'] = ''
    form.to_csv(form_path, index=False)
print('Форма для своих ответов:', form_path)
annotations = pd.read_csv(DATA_DIR / 'annotation' / 'example_annotations.csv')
display(annotations.head())
```

Теперь соединим A и B по ключу. `validate='one_to_one'` в `merge` запрещает незаметное размножение строк
из-за повторяющихся ключей. `how='outer'` оставляет ключи обеих таблиц: так видны отсутствующие ответы.


```python
a = annotations.loc[annotations['annotator'] == 'example_A', ['item_id', 'label']]
b = annotations.loc[annotations['annotator'] == 'example_B', ['item_id', 'label']]
a = a.rename(columns={'label': 'label_a'})
b = b.rename(columns={'label': 'label_b'})
joined = a.merge(b, on='item_id', how='outer', validate='one_to_one')
known = joined['label_a'].notna() & joined['label_b'].notna()
pairs = joined.loc[known].copy()
missing = joined.loc[~known].copy()
print('Пар для сравнения:', len(pairs), 'Без полного ответа:', len(missing))
display(missing)
```

## 3. Процент совпадений — начало, а не весь отчёт

Для каждой пары проверяем равенство меток. Среднее массива True/False — доля совпадений:
True считается единицей, False — нулём. Знаменатель — **29 полных пар**, не все 30 исходных строк.
Сохраним также сами разногласия: одна цифра не расскажет, какой пункт инструкции надо исправить.


```python
same = pairs['label_a'] == pairs['label_b']
observed = same.mean()
disagreements = pairs.loc[~same].merge(messages[['item_id', 'text']], on='item_id')
print('Совпало:', int(same.sum()), 'из', len(pairs), 'Доля:', round(observed, 3))
display(disagreements)
disagreements.to_csv(output_dir / 'example_disagreements.csv', index=False)
```

## 4. А если почти все ответы одной категории?

Если A и B почти всегда выбирают issue, они часто совпадут даже без согласованного чтения текста.
Каппа Коэна сравнивает наблюдаемое совпадение $p_o$ с ориентиром $p_e$: совпадением при независимом
выборе меток с теми же частотами у каждого разметчика.

Пример: A выбирает issue в 3/4 случаев, B — в 1/2. Независимое совпадение issue даёт
$3/4\cdot1/2$, второй категории — $1/4\cdot1/2$. Сумма $p_e=0.5$.
Если фактически совпали 3/4 ответов, $\kappa=(0.75-0.5)/(1-0.5)=0.5$.
Это не «50% правильных ответов»: истинных меток мы не знаем.

`value_counts(normalize=True)` даёт доли категорий. `reindex` добавит отсутствующие категории с нулём.
Используем один порядок меток, иначе можно умножить issue на question.


```python
labels = ['issue', 'request', 'question', 'unclear']
freq_a = pairs['label_a'].value_counts(normalize=True).reindex(labels, fill_value=0)
freq_b = pairs['label_b'].value_counts(normalize=True).reindex(labels, fill_value=0)
expected = (freq_a * freq_b).sum()
kappa = (observed - expected) / (1 - expected)
display(pd.DataFrame({'A': freq_a, 'B': freq_b, 'product': freq_a * freq_b}))
print('Наблюдаемое:', observed, 'Ориентир:', expected, 'Каппа:', kappa)
```

Каппа 1 — полное согласие при невырожденных частотах; около 0 — на уровне данного ориентира;
отрицательная — ниже него. Универсальный проходной порог не назначаем: смотрим задачу, частоты,
число примеров и последствия ошибок. Согласованная неверная инструкция может дать отличную каппу.
При сильном перекосе категорий каппа и доля совпадений могут производить разное впечатление;
в отчёте нужны **обе**, распределения, размер выборки и примеры разногласий.

## Резерв: а если разметчиков трое?

Каппа Флейса в простом варианте требует одинакового числа ответов на каждый объект. Здесь только
учебный пример из двух категорий: строки — объекты, столбцы — число голосов за категории.
Для [2,1] из 3 ответов согласованные **упорядоченные** пары: $2\cdot1+1\cdot0=2$ из $3\cdot2=6$.
Среднее такого согласия по объектам сравнивается с суммой квадратов общих долей категорий.
Это не автоматическое обобщение любой неполной таблицы: пропуски и разное число ответов требуют
другой обработки. Не подменяем ранги экспертов категориями.


```python
votes = np.array([[3,0], [2,1], [1,2], [0,3]])
n_raters = votes.sum(axis=1)
assert (n_raters == n_raters[0]).all()
per_item = (votes * (votes - 1)).sum(axis=1) / (n_raters * (n_raters - 1))
p_bar = per_item.mean()
category_share = votes.sum(axis=0) / votes.sum()
p_chance = (category_share ** 2).sum()
fleiss = (p_bar - p_chance) / (1 - p_chance)
print('Согласие по объектам:', per_item, 'Каппа Флейса:', fleiss)
```
