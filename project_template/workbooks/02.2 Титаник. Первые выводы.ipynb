### Пропуск — это тоже информация

`None` обозначает отсутствие значения. В числовом столбце pandas пропуск часто показывается как NaN; в новых типах может быть `<NA>`. Из-за пропуска целочисленный столбец может стать дробным. Не будем проверять пропуск сравнением `== None`: используем `.isna()`.


```python
demo = pd.Series([10, 20, None])
print(demo)
print(demo.isna())
```

isna вернул маску True/False. При суммировании True считается единицей, False — нулём. Поэтому сумма маски — число пропусков. Сначала убедимся на маленьком примере.

Метод `.sum()` складывает элементы Series: для маски это число True.


```python
missing_demo = demo.isna()
print(missing_demo.sum())
```

У DataFrame метод isna возвращает целую таблицу масок. sum по умолчанию складывает по строкам внутри каждого столбца: получим количество пропусков в каждом поле. Не удаляем строки автоматически: сначала выясняем смысл отсутствия.


```python
missing_mask = df.isna()
missing_counts = missing_mask.sum()
missing_counts
```

`.idxmax()` возвращает **метку** максимального элемента Series, а `.max()` — само значение. Их легко перепутать. Здесь метки — названия столбцов.


```python
print('Название:', missing_counts.idxmax())
print('Количество:', missing_counts.max())
```

Классы билета можно посчитать через `.value_counts()`. Метод считает, сколько раз встретилось каждое значение. Результат по умолчанию упорядочен по частоте, а нам удобнее классы 1,2,3 — применим `.sort_index()` к полученной Series.


```python
classes = df['Pclass']
counts = classes.value_counts()
display(counts)
ordered_counts = counts.sort_index()
display(ordered_counts)
```

## Хочется увидеть конкретных пассажиров

Начнём с одного столбца возраста. Сравнение Series с 18 выполняется для каждого значения отдельно и возвращает маску. Для пропусков результат может отображаться как False или NA в зависимости от типа; loc не включит такие строки. «Не отобрали» не означает «доказали, что это ребёнок».


```python
ages = df['Age']
adult_mask = ages >= 18
adult_mask.head()
```

Передадим маску в loc: строки с True останутся. Второй аргумент после запятой — список нужных столбцов. Сначала выберем все столбцы, затем сузим вывод: никаких новых условий пока не добавляем.


```python
adults = df.loc[adult_mask]
adult_details = adults[['Name', 'Age', 'Fare']]
adult_details.head()
```

Добавим класс билета. Второе условие отдельно проверит Pclass==3. Оператор `&` объединяет маски по правилу «и», `|` — «или». Обычное `and` не подходит для Series. Если записывать сравнения сразу вместе, каждое нужно заключить в скобки.


```python
third_class_mask = df['Pclass'] == 3
both_masks = adult_mask & third_class_mask
adult_third = df.loc[both_masks, ['Name', 'Age', 'Fare']]
adult_third.head()
```

Теперь хотим дорогие билеты первыми. `sort_values('Fare', ascending=False)` сортирует строки по Fare в убывающем порядке. `ascending=True` был бы возрастающим. Исходный df не изменяется: результат сохраняем отдельно. Последующий head ограничивает только показ.


```python
by_fare = adult_third.sort_values('Fare', ascending=False)
by_fare.head()
```

## Среднее у нулей и единиц

Для списка `[1,0,1,1]` сумма равна 3 — столько успешных исходов. Делим на 4, получаем долю 0.75. Поэтому `.mean()` у нулей и единиц равен доле единиц, а не «среднему человеку».


```python
outcomes = pd.Series([1, 0, 1, 1])
print(outcomes.sum())
print(outcomes.mean())
```

У Survived такой же смысл. Сначала посчитаем по всей выборке, затем отдельно по одному классу через уже знакомую маску.


```python
print(df['Survived'].mean())
first_class_mask = df['Pclass'] == 1
first_class = df.loc[first_class_mask]
print(first_class['Survived'].mean())
```

Хотим повторить это для каждого класса. `.groupby('Pclass')` объединяет строки по значению Pclass. Это ещё не результат расчёта, а способ группировки. Затем выбираем столбец Survived и действие mean. Разнесём три шага по именам, чтобы не скрыть последовательность.


```python
groups = df.groupby('Pclass')
group_outcomes = groups['Survived']
rates_by_class = group_outcomes.mean()
rates_by_class
```

Доля без размера группы может вводить в заблуждение. `.count()` считает непустые значения в каждой группе. В Survived пропусков нет, поэтому здесь это количество наблюдений. Не путайте count с sum: сумма нулей и единиц считает только выживших.


```python
counts_by_class = group_outcomes.count()
counts_by_class
```

## Пара слов о признаках

Когда мы изучаем датафрейм с целью анализа данных, то стоит понимать, какие признаки встречаются в столбцах. Их можно разделить на следующие группы:

1. Количественные
   1. Дискретные
   2. Непрерывные
2. Качественные (категориальные)
   1. Порядковые
   2. Номинальные

В чем их отличие?

Количественные признаки задаются числом. При этом дискретные задаются конкретными числами с неким шагом. Например, число студентов в аудитории мы можем описать только целыми неотрицательными числами. У нас не может сидеть полтора студента. А вот непрерывные признаки можно задать любым числом. Например, масса камня или скорость автомобиля.

Качественные признаки могут быть заданы текстово, либо числами, если отражают разбиение на категории. Например, наличие детей в семье, заданное числами 0 и 1 будет не количественным, а качественным признаком, хоть и указано числами. Порядковые признаки отражают некий порядок. Например, уровень образования: начальное, общее, среднее и т.д. А номинальные нельзя распределить по порядку. Примером этих признаков могут быть цвета воздушных шариков: красный. синий, зеленый...
