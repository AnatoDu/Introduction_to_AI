Теперь нам нужен только текст заголовка. Метод `.get_text()` снимает разметку. Не будем писать оба действия в одной строке: так пока проще понимать, где мы находимся.

````python
price_tag = tiny_soup.find('p', class_='price')
print(price_tag)
price_text = price_tag.get_text()
print(repr(price_text))
````

`repr` показывает строку так, чтобы были видны пробелы по краям и специальные символы. Метод строки `.strip()` убирает пробелы по краям, но не внутри. Хотели цену — получили ещё и пробелы, теперь уточнили обработку. Вот и весь маленький шаг.


````python
neat_price = price_text.strip()
print(repr(neat_price))
````

### Хотим то же для любой карточки

Значит, пора собрать знакомые строки в функцию. `def record_from_card(card)` объявляет действие и его вход. `return record` возвращает результат вызывающему коду. До вызова тело функции не выполняется. Имена внутри — местные переменные: `card` при каждом вызове будет новым входом.

Код ниже длиннее предыдущих ячеек, потому что мы **собираем уже разобранные части**. Найдите в нём поиск тега, извлечение текста и словарь. Нового способа поиска здесь нет. Функция рассчитана на структуру нашего учебного каталога; если поле исчезнет, надо разбираться в изменившейся странице, а не объявлять пустой результат успехом.


````python
def record_from_card(card):
    title_tag = card.find('a', attrs={'data-field': 'title'})
    address_tag = card.find('p', attrs={'data-field': 'address'})
    price_tag = card.find('p', attrs={'data-field': 'price'})
    area_tag = card.find('p', attrs={'data-field': 'area'})
    record = {
        'listing_id': card['data-id'],
        'title': title_tag.get_text(' ', strip=True),
        'address': address_tag.get_text(' ', strip=True),
        'price': price_tag.get_text(' ', strip=True),
        'area': area_tag.get_text(' ', strip=True),
        'href': title_tag['href'],
    }
    return record

second_record = record_from_card(cards[1])
second_record
````

## По желанию: если страница в интернете

До сих пор источник строки — локальный файл. `requests.get(адрес)` получает ответ сервера: объект Response. Сам ответ, его код и его текст — разные вещи. `timeout=(3,5)` задаёт ожидание соединения и чтения; это не гарантия полного времени запроса. `raise_for_status()` сообщает об HTTP-ошибках. Даже код 200 не доказывает, что в тексте нужный каталог.

Ниже **образец для отдельной ячейки**, автоматически он не выполняется. Сайт предназначен для упражнений; его структура другая. При недоступной сети продолжаем локально. Сначала ознакомьтесь с правилами источника; не обходите ограничения или капчу.

```python
import requests
response = requests.get('https://books.toscrape.com/', timeout=(3, 5))
print(response.status_code)
response.raise_for_status()
web_html = response.text
print(web_html[:300])
```

Не объединяйте все четыре шага в одну строку, пока не сможете объяснить каждый. Справка: [Requests](https://requests.readthedocs.io/en/latest/user/quickstart/).
