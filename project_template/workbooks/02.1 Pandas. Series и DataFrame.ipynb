### Новые структуры хранения данных и их создание

Pandas открывает доступ к замечательным структурам - Series и DataFrame.

- Series - это одномерный массив с метками индекса. Похож на словарь, но метки могут повторяться.
- DataFrame - двумерная таблица, где каждый столбец представляет собой Series.

Но, давайте лучше посмотрим, как они выглядят на практике. Создадим Series.

```python
fruits = pd.Series(['apple', 'orange', 'mango'])
fruits
```

Смотрите, у нас сами собой появились индексы. По умолчанию это числовой ряд с нуля, однако его можно задать вручную.

```python
code = [11, 43, 77]
region = ['Komi', 'Kirov', 'Moscow']

codes = pd.Series(data=region, index=code)
codes
```

Теперь и индексы можно задавать самостоятельно. И они не обязательно должны быть числами.

```python
codes_2 = pd.Series(data=code, index=region)
codes_2
```

По индексу мы всегда можем обратиться и получить значение в данной строке серии.

```python
codes[11]
```

```python
codes_2['Kirov']
```

Это короткая запись обращения. А теперь попробуем loc — доступ по метке. С таблицами он умеет ещё кое-что полезное, скоро посмотрим.

```python
codes.loc[11]
```

А вот если мы хотим какую-либо строку по порядку, то используем iloc (нумерация с нуля).

```python
codes.iloc[1]
```

Для срезов по позиции используйте `iloc`: правая граница не включается. В срезе `loc` обе граничные метки включаются, если присутствуют. Предскажите результат перед запуском.

Запись среза — начало:конец:шаг. Если начало пропущено, берём с начала; если конец пропущен — до конца. Шаг по умолчанию 1. Отрицательная позиция отсчитывается с конца: −1 — последний элемент. Давайте сначала предсказывать каждую строку, затем запускать её и сверять ответ.


```python
codes.iloc[0:2]
```

```python
codes.iloc[:2]
```

```python
codes.iloc[1:]
```

```python
codes.iloc[1:1]
# Последний не войдёт
```

```python
codes.iloc[-2:]
# Да, срезы можно делать с конца именно так
```

```python
codes.iloc[:6]
```

```python
codes.iloc[::2]
# А третье число отвечает за частоту выборки элементов, тут каждый второй
```

Мы умеем выбирать значения, теперь попробуем сложить их. Метод `.sum()` вычисляет сумму Series. В скобках ничего не передаём: суммируются все значения. `print` печатает результат. Для упражнения индекс можно задать обычным списком строк — цикл здесь не нужен.


```python
numbers = pd.Series([10, 20, 30])
print(numbers.sum())
```

В `values` каждый внутренний список — одна строка оценок по предмету. `index=subjects` даст имена строкам, `columns=students` — столбцам. Угадайте: сколько строк и сколько столбцов получится?


```python
students = ['Ivanov', 'Petrov', 'Sidorov']
subjects = ['Algebra', 'Geometry', 'Chemistry', 'Physics', 'History', 'Biology', 'Literature']
values = [[5, 4, 3], [5, 3, 3], [4, 5, 3], [4, 4, 3], [4, 5, 3], [4, 5, 3], [4, 4, 5]]

marks = pd.DataFrame(data=values, index=subjects, columns=students)
display(marks)
```

Как видите, тут мы можем задать ещё названия столбцов. Обратите внимание на то, как двумерный список преобразуется в таблицу DataFrame, это происходит построчно. Это важно помнить, чтобы не создать случайно транспонированную таблицу.

Что за display? Да, можно было обойтись и без него, ведь последнюю строку Jupyter и так бы показал, но запомните эту функцию. Делает она то же самое, что и print, но фреймы выводит более красиво. Сравните со следующей ячейкой.

В следующем примере ошибка намеренная. `try` пробует действие, `except KeyError as error` перехватывает именно ошибку отсутствующей метки и печатает её текст. Так тетрадь сможет продолжить выполнение. Это демонстрация различия строки и столбца, а не рекомендуемый способ скрывать любые ошибки.


А теперь вновь пару слов об обращении к частям фрейма. В серии при простом обращении в квадратных скобках мы могли получить значение интересующей строки, во фреймах строку вы так не получите, только ошибку, а вот столбец - без проблем. 

```python
try:
    marks['Algebra']
except KeyError as error:
    print("Ожидаемая ошибка выбора столбца:", error)
```

```python
marks['Petrov']
```

А если требуются несколько столбцов, то мы не пишем несколько названий, а передаем список из этих столбцов, иначе получите ошибку.

```python
try:
    marks['Petrov', 'Sidorov']
except KeyError as error:
    print("Ожидаемая ошибка выбора столбца:", error)
```

```python
marks[['Petrov', 'Sidorov']]
```

Тут самое время вспомнить про loc, он тут работает замечательно (iloc тоже).

```python
marks.loc['Algebra']
```

А теперь можно сказать про полную форму этого метода, ведь в нём можно указать и индексы, и столбцы.

```python
marks.loc['Algebra', 'Petrov']
```

У iloc тоже есть два места в квадратных скобках: **позиция строки, позиция столбца**. Считаем с нуля. Попробуем Geometry у Sidorov: строка 1, столбец 2. В следующем задании самостоятельно найдите позиции другой оценки.


```python
print(marks.iloc[1, 2])
print(marks.loc['Geometry', 'Sidorov'])
```
