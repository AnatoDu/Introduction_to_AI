- **назначение** — какую задачу решает;
- **сфера применения** — где её используют;
- **виды знаний** — откуда она «знает», что делать: из правил, которые записали люди, из данных, на которых её обучили, или из того и другого;
- **временные аспекты** — отвечает мгновенно или может обработать всё ночью.

Вот как это выглядит на примере антиспам-фильтра.
